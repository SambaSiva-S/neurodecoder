```python
# ===== NEURODECODER: MILESTONE 1 — SIGNAL EXPLORATION =====
import numpy as np
import scipy.signal as sig
import matplotlib.pyplot as plt
import mne
from braindecode.datasets import MOABBDataset

print("Loading BCI Competition IV-2a dataset...")
dataset = MOABBDataset(dataset_name="BNCI2014_001", subject_ids=[1])
raw = dataset.datasets[0].raw
sfreq = raw.info['sfreq']
print(f"Channels: {len(raw.ch_names)}, Sampling rate: {sfreq} Hz, Duration: {raw.times[-1]:.1f}s")
print("Dataset loaded!\n")

# ===== 1. RAW EEG VISUALIZATION =====
channels_to_plot = ['C3', 'Cz', 'C4']
data = raw.get_data(picks=channels_to_plot)

fig, axes = plt.subplots(3, 1, figsize=(14, 8), sharex=True)
t_10s = np.arange(int(10*sfreq)) / sfreq
for i, ch_name in enumerate(channels_to_plot):
    axes[i].plot(t_10s, data[i, :int(10*sfreq)] * 1e6, linewidth=0.5)
    axes[i].set_ylabel(f'{ch_name}\n(µV)')
    axes[i].grid(True, alpha=0.3)
axes[0].set_title('Raw EEG - Subject 1 - Motor Cortex Channels', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Time (seconds)')
plt.tight_layout()
plt.show()
print("C3 = left hemisphere, C4 = right hemisphere, Cz = center\n")

# ===== 2. POWER SPECTRAL DENSITY =====
fig, ax = plt.subplots(figsize=(12, 5))
for ch_name in ['C3', 'C4']:
    ch_data = raw.get_data(picks=[ch_name])[0]
    freqs, psd = sig.welch(ch_data, fs=sfreq, nperseg=int(2*sfreq))
    ax.semilogy(freqs, psd * 1e12, label=ch_name, linewidth=2)
ax.axvspan(8, 12, alpha=0.2, color='orange', label='Mu (8-12 Hz)')
ax.axvspan(13, 30, alpha=0.2, color='green', label='Beta (13-30 Hz)')
ax.set_xlim([0, 50])
ax.set_xlabel('Frequency (Hz)')
ax.set_ylabel('Power (µV²/Hz)')
ax.set_title('Power Spectral Density - Motor Cortex', fontsize=14, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ===== 3. BANDPASS FILTERING =====
bands = {
    'Delta (0.5-4 Hz)': (0.5, 4),
    'Theta (4-8 Hz)': (4, 8),
    'Mu (8-12 Hz)': (8, 12),
    'Beta (13-30 Hz)': (13, 30),
    'Gamma (30-45 Hz)': (30, 45)
}

c3_data = raw.get_data(picks=['C3'])[0]
fig, axes = plt.subplots(len(bands)+1, 1, figsize=(14, 16), sharex=True)
axes[0].plot(t_10s, c3_data[:int(10*sfreq)] * 1e6, linewidth=0.5, color='black')
axes[0].set_ylabel('Raw\n(µV)')
axes[0].set_title('C3 — Raw EEG vs Filtered Frequency Bands', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

colors = ['purple', 'blue', 'orange', 'green', 'red']
for i, (band_name, (low, high)) in enumerate(bands.items()):
    b, a = sig.butter(4, [low/(sfreq/2), high/(sfreq/2)], btype='band')
    filtered = sig.filtfilt(b, a, c3_data)
    axes[i+1].plot(t_10s, filtered[:int(10*sfreq)] * 1e6, linewidth=0.5, color=colors[i])
    axes[i+1].set_ylabel(f'{band_name}\n(µV)')
    axes[i+1].grid(True, alpha=0.3)
axes[-1].set_xlabel('Time (seconds)')
plt.tight_layout()
plt.show()
print("Mu (orange) = the motor rhythm your decoder will detect\n")

# ===== 4. SPECTROGRAM =====
fig, axes = plt.subplots(2, 1, figsize=(14, 8))
for i, ch_name in enumerate(['C3', 'C4']):
    ch_data = raw.get_data(picks=[ch_name])[0]
    f, t_spec, Sxx = sig.spectrogram(ch_data[:int(60*sfreq)], fs=sfreq, nperseg=int(sfreq), noverlap=int(sfreq*0.9))
    axes[i].pcolormesh(t_spec, f, 10*np.log10(Sxx*1e12), shading='gouraud', cmap='viridis', vmin=-20, vmax=20)
    axes[i].set_ylim([0, 45])
    axes[i].set_ylabel('Frequency (Hz)')
    axes[i].set_title(f'{ch_name} — Spectrogram (first 60s)', fontsize=13)
    axes[i].axhline(y=8, color='white', linestyle='--', alpha=0.5)
    axes[i].axhline(y=12, color='white', linestyle='--', alpha=0.5)
axes[1].set_xlabel('Time (seconds)')
fig.suptitle('Time-Frequency Analysis — Motor Cortex', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("White dashed lines = Mu band boundaries\n")

# ===== 5. ERD/ERS — THE BCI SIGNAL =====
events, event_id = mne.events_from_annotations(raw)
print(f"Event types: {event_id}")

# Create epochs with only C3 and C4 for ERD analysis
epochs_motor = mne.Epochs(raw, events, event_id, tmin=-1, tmax=4.0,
                    picks=['C3', 'C4'],
                    baseline=None, preload=True, verbose=False)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for col, class_name in enumerate(['left_hand', 'right_hand']):
    if class_name not in epochs_motor.event_id:
        continue
    for row, ch_name in enumerate(['C3', 'C4']):
        class_epochs = epochs_motor[class_name].get_data(picks=[ch_name])[:, 0, :]
        
        mu_power_trials = []
        for trial in class_epochs:
            b, a = sig.butter(4, [8/(sfreq/2), 12/(sfreq/2)], btype='band')
            mu_signal = sig.filtfilt(b, a, trial)
            power = mu_signal ** 2
            kernel_size = int(0.25 * sfreq)
            smoothed = np.convolve(power, np.ones(kernel_size)/kernel_size, mode='same')
            mu_power_trials.append(smoothed)
        
        avg_power = np.mean(mu_power_trials, axis=0)
        times_epoch = np.linspace(-1, 4, len(avg_power))
        
        baseline_idx = times_epoch < 0
        baseline_power = np.mean(avg_power[baseline_idx])
        erd_percent = ((avg_power - baseline_power) / baseline_power) * 100
        
        axes[row, col].plot(times_epoch, erd_percent, linewidth=2, color='blue')
        axes[row, col].axhline(y=0, color='black', linestyle='--', alpha=0.5)
        axes[row, col].axvline(x=0, color='red', linestyle='--', alpha=0.5, label='Cue onset')
        axes[row, col].fill_between(times_epoch, erd_percent, 0,
                                     where=(erd_percent < 0), alpha=0.3, color='blue', label='ERD')
        axes[row, col].fill_between(times_epoch, erd_percent, 0,
                                     where=(erd_percent > 0), alpha=0.3, color='red', label='ERS')
        axes[row, col].set_ylabel('Power change (%)')
        axes[row, col].set_title(f'{ch_name} during {class_name.upper()}', fontsize=12)
        axes[row, col].set_ylim([-60, 60])
        axes[row, col].grid(True, alpha=0.3)
        axes[row, col].legend(fontsize=8)

axes[1, 0].set_xlabel('Time (seconds)')
axes[1, 1].set_xlabel('Time (seconds)')
fig.suptitle('ERD/ERS — Mu Band (8-12 Hz)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("RIGHT hand imagery → C3 should show ERD (blue dip)")
print("LEFT hand imagery → C4 should show ERD (blue dip)")
print("Contralateral pattern = what your decoder detects\n")

# ===== 6. TOPOGRAPHIC MAP =====
# Create separate epochs with ALL EEG channels for topographic maps
eeg_picks = mne.pick_types(raw.info, eeg=True, eog=False, stim=False)
epochs_topo = mne.Epochs(raw, events, event_id, tmin=0.5, tmax=4.0,
                         picks=eeg_picks, baseline=None, preload=True, verbose=False)

info_eeg = epochs_topo.info

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
class_names = [k for k in epochs_topo.event_id.keys()]

for idx, class_name in enumerate(class_names[:4]):
    class_data = epochs_topo[class_name].get_data()
    
    mu_powers = []
    for ch_idx in range(class_data.shape[1]):
        ch_trials = class_data[:, ch_idx, :]
        powers = []
        for trial in ch_trials:
            b, a = sig.butter(4, [8/(sfreq/2), 12/(sfreq/2)], btype='band')
            mu_sig = sig.filtfilt(b, a, trial)
            powers.append(np.mean(mu_sig ** 2))
        mu_powers.append(np.mean(powers))
    
    mu_powers = np.array(mu_powers)
    mne.viz.plot_topomap(mu_powers, info_eeg, axes=axes[idx], show=False,
                          cmap='RdBu_r')
    axes[idx].set_title(f'{class_name}', fontsize=12)

fig.suptitle('Mu Band Power Distribution — By Movement Class', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("Each head = Mu power during different imagined movements\n")

# ===== 7. ALL SUBJECTS COMPARISON =====
print("Loading all 9 subjects...")
subject_erd = {}

for subj_id in range(1, 10):
    try:
        ds = MOABBDataset(dataset_name="BNCI2014_001", subject_ids=[subj_id])
        subj_raw = ds.datasets[0].raw
        subj_sfreq = subj_raw.info['sfreq']
        subj_events, subj_event_id = mne.events_from_annotations(subj_raw)
        subj_epochs = mne.Epochs(subj_raw, subj_events, subj_event_id,
                                  tmin=0.5, tmax=2.5, picks=['C3', 'C4'],
                                  baseline=None, preload=True, verbose=False)
        
        if 'left_hand' in subj_event_id and 'right_hand' in subj_event_id:
            left_data = subj_epochs['left_hand'].get_data(picks=['C3'])[:, 0, :]
            right_data = subj_epochs['right_hand'].get_data(picks=['C3'])[:, 0, :]
            
            b, a = sig.butter(4, [8/(subj_sfreq/2), 12/(subj_sfreq/2)], btype='band')
            left_mu = np.mean([np.mean(sig.filtfilt(b, a, trial)**2) for trial in left_data])
            right_mu = np.mean([np.mean(sig.filtfilt(b, a, trial)**2) for trial in right_data])
            
            erd_diff = abs(left_mu - right_mu) / max(left_mu, right_mu) * 100
            subject_erd[f'S{subj_id}'] = erd_diff
            print(f"  Subject {subj_id}: ERD difference = {erd_diff:.1f}%")
    except Exception as e:
        print(f"  Subject {subj_id}: Error - {e}")

fig, ax = plt.subplots(figsize=(10, 5))
subjects = list(subject_erd.keys())
values = list(subject_erd.values())
colors_bar = ['green' if v > 10 else 'orange' if v > 5 else 'red' for v in values]
ax.bar(subjects, values, color=colors_bar, edgecolor='black')
ax.set_ylabel('ERD Difference (%)')
ax.set_xlabel('Subject')
ax.set_title('Signal Quality Across 9 Subjects (Higher = Easier to Decode)', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

print("\nGreen = strong signal, Orange = moderate, Red = weak (BCI illiteracy)")

print("\n" + "="*60)
print("MILESTONE 1 COMPLETE!")
print("="*60)
print("\nWhat you built:")
print("  1. Raw EEG visualization")
print("  2. Power Spectral Density")
print("  3. Bandpass filtering (5 frequency bands)")
print("  4. Time-frequency spectrograms")
print("  5. ERD/ERS quantification")
print("  6. Topographic scalp maps")
print("  7. Cross-subject signal quality comparison")
print("\nNext: Milestone 2 — Classical Decoder (CSP + LDA/SVM)")
```
